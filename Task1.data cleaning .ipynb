```python
# DATA CLEANING AND PROCESSING

# data set
# 1. Dataset Imported using pandas
import pandas as pd

data = {
    "Customer_ID": [101, 102, 103, 104, 105, 105],
    "Name": ["Amit", "Priya", "Rahul", "Sneha", "Neha", "Neha"],
    "Age": [25, 30, None, 28, 35, 35],
    "City": ["Pune", "Mumbai", "Pune", "Pune", None, "Pune"],
    "Purchase_Amount": [5000, 7500, 3000, None, 9000, 9000]
}

df = pd.DataFrame(data)

df
```

```python
# 2. Dataset Structure Inspected
# dataset  basic information

df.info()
```

```python
df.shape
```

```python
df.head()
```

```python
# 3. missing values Identified

df.isnull().sum()
```

```python
# 4. Duplicates Records Identified
df.duplicated().sum()
```

```python
df.isnull().sum()
```

```python
# 6.Missing values Handled

# missing value age 
df["Age"]=df["Age"].fillna(df["Age"].median())
df
```

```python
# missing value city

df["City"] = df["City"].fillna(df["City"].mode()[0])
df
```

```python
# Purchase Amount

df["Purchase_Amount"] = df["Purchase_Amount"].fillna(df["Purchase_Amount"].median())
df
```

```python
# check missing vale
df.isnull().sum()
```

```python
# 7.Data Types checked

df.dtypes
```

```python
# check data

df.info()
```

```python
# 8. cleaned dataset prepared
df.isnull().sum()
```

```python
# 9. cleaned CSV Explorted

df.to_csv("cleaned_customer_sales.csv", index=False)
```
