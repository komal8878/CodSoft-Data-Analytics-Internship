```python
# Data visualization 

## objective
the objectives of this task is to create visualizations using python, pandas, matplotlib, and Seaborn to identify
patterns,treads, and relationships in the dataset.
```

```python
# import libraries

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")
```

```python
## 1. load the Dataset

df= pd.read_csv("advertising.csv")

print("Dataset loaded successfully!")
display(df.head())
```

```python
## 2. dataset Overview

print("Dataset shape:", df.shape)

print("\nColumn Names:")
print(df.columns.tolist())

print("\nDataset Information:")
df.info()
```

```python
## 3. Bar Chart

df[["TV", "radio", "newspaper"]].mean().plot(
    kind="bar",
    figsize=(8, 5)
)

plt.title("Average Advertising Spending by Media")
plt.xlabel("Advertising Media")
plt.ylabel("Average Spending")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
print(type(df))
print(df.head())
```

```python
## 4. Scatter Plot

plt.figure(figsize=(8, 5))

plt.scatter(df["TV"], df["sales"])

plt.title("TV Advertising vs Sales")
plt.xlabel("TV Advertising")
plt.ylabel("Sales")

plt.tight_layout()
plt.show()
```

```python
## 5. Histogram

plt.figure(figsize=(8, 5))

plt.hist(df["sales"], bins=15, edgecolor="black")

plt.title("Distribution of Sales")
plt.xlabel("Sales")
plt.ylabel("Frequency")

plt.tight_layout()
plt.show()
```

```python
## 6. Correlation Heatmap

plt.figure(figsize=(8, 6))

sns.heatmap(
    df[["TV", "radio", "newspaper", "sales"]].corr(),
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Correlation Heatmap")

plt.tight_layout()
plt.show()
```

```python
## 7. Line Chart
```

```python
plt.figure(figsize=(8, 5))

plt.plot(df.index, df["sales"])

plt.title("Sales Across Dataset Records")
plt.xlabel("Record Number")
plt.ylabel("Sales")

plt.tight_layout()
plt.show()
```

```python
## 8. Conclusion

Data visualization was performed using Python, Pandas,
Matplotlib, and Seaborn.

Bar charts, scatter plots, histograms, line charts,
and a correlation heatmap were created to understand
the dataset.

These visualizations help identify data distributions,
patterns, and relationships between advertising variables
and sales.

The analysis demonstrates how data visualization can
support data interpretation and data-driven decision-making.
```
